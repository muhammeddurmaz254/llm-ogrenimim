```python
#system prompt is the instruction to the assistant on how to summarize the content of a website
system_prompt = """You are an assistant that summarizes the content of a website.
You will be given the content of a website, and you will summarize it in a few sentences. 
ignore any content that is not relevant to the main topic of the website.
Respond in markdown."""

#user prompt is the content of the website that will be summarized
user_prompt= """ 
Here is the content of a website:
Provide a short summary of the content in a few sentences.
"""
```

```python
def summarize_website_content(system_prompt, user_prompt, url):

    client = OpenAI(
        api_key=os.environ.get("GEMINI_API_KEY"),
        base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
    )
    url_content = fetch_website_contents(url)

    response = client.chat.completions.create(
        model="gemini-3.6-flash",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt + url_content}
        ]
    )

    print(response.choices[0].message) 
```

```python
summarize_website_content(system_prompt, user_prompt, url="https://www.haberturk.com/ozel-icerikler/murat-bardakci/3904745-eyym-i-bhur")
```
